# Named Entity Recognition (NER)

Extract **people, organizations, and places** from text with NLTK.

**Use case:** Contract parsing, news monitoring, CRM auto-fill. Production systems often use spaCy.

**Prerequisites:** `01-nlp-fundamentals.ipynb` introduces the concepts. This notebook uses `nlp_helpers.py` for preprocessing.


```python
import numpy as np
import pandas as pd

import nltk
from nltk.tokenize import word_tokenize

from nlp_helpers import download_nltk_data, preprocess_text

download_nltk_data(['maxent_ne_chunker_tab', 'words'])
print('Setup complete.')
```

## 1. Extract entities

`ne_chunk` groups tokens into subtrees labeled PERSON, ORGANIZATION, GPE (geo-political), etc.


```python
def extract_entities(text):
    tokens = word_tokenize(text)
    pos_tags = nltk.pos_tag(tokens)
    chunks = nltk.ne_chunk(pos_tags)
    entities = []
    for chunk in chunks:
        if hasattr(chunk, 'label'):
            entities.append((chunk.label(), ' '.join(c[0] for c in chunk)))
    return entities

sample = "Apple Inc. announced a deal with Microsoft in San Francisco on January 15, 2024."
entities = extract_entities(sample)
print("Text:", sample)
print("\nExtracted entities:")
for label, entity in entities:
    print(f"  {label}: {entity}")
```
