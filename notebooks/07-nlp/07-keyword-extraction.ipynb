# Keyword Extraction from Feedback

Surface the most distinctive terms in a single document using **TF-IDF**.

**Use case:** Auto-tag tickets, summarize surveys, highlight UX pain points.

**Prerequisites:** `01-nlp-fundamentals.ipynb` introduces the concepts. This notebook uses `nlp_helpers.py` for preprocessing.


```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer

from nlp_helpers import download_nltk_data, preprocess_text

download_nltk_data()
print('Setup complete.')
```

## 1. TF-IDF keywords

We compare the document to generic filler sentences so **IDF** down-weights words that appear everywhere. High TF-IDF terms are strong keyword candidates for that specific document.

```python
def extract_keywords(text, top_n=5):
    """Extract top keywords from a single document using TF-IDF (vs a small corpus)."""
    # Compare doc against itself + a few generic docs to get IDF effect
    corpus = [text, "generic filler document for idf", "another generic document"]
    vec = TfidfVectorizer(stop_words='english')
    X = vec.fit_transform(corpus)
    scores = X[0].toarray().flatten()
    words = vec.get_feature_names_out()
    top_idx = np.argsort(scores)[-top_n:][::-1]
    return [(words[i], scores[i]) for i in top_idx if scores[i] > 0]

feedback = """The onboarding process was confusing. The dashboard is cluttered and hard to navigate.
Customer support was helpful when I called. Overall the product works but needs better UX."""

print("Feedback:")
print(feedback)
print("\nExtracted keywords:")
for word, score in extract_keywords(feedback):
    print(f"  {word}: {score:.3f}")
```
