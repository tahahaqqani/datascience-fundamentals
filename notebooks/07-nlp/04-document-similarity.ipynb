# Document Similarity & Search

Rank knowledge-base articles by **cosine similarity** to a user query.

**Use case:** FAQ search, similar-ticket lookup, content recommendations.

**Prerequisites:** `01-nlp-fundamentals.ipynb` introduces the concepts. This notebook uses `nlp_helpers.py` for preprocessing.


```python
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from nlp_helpers import download_nltk_data, preprocess_text

download_nltk_data()
print('Setup complete.')
```

## 1. Build a small knowledge base

Each string is a help article. In production this would be thousands of docs.


```python
# Document corpus (e.g., product descriptions or knowledge base articles)
docs = [
    "How to reset your password and recover your account",
    "Troubleshooting login issues and authentication errors",
    "Refund policy and how to request a refund",
    "Shipping options and delivery times",
    "Returning items and exchange process"
]

vec = TfidfVectorizer()
X_docs = vec.fit_transform(docs)

# Query: user search
query = "I forgot my password and need to get back into my account"
query_vec = vec.transform([preprocess_text(query)])

# Cosine similarity
sims = cosine_similarity(query_vec, X_docs)[0]

print("Query:", query)
print("\nSimilar documents (ranked by similarity):")
for i, idx in enumerate(np.argsort(sims)[::-1]):
    print(f"  {i+1}. [{sims[idx]:.3f}] {docs[idx]}")
```

## How it works

1. Fit TF-IDF on the document corpus.
2. Transform the query with the **same** vectorizer.
3. `cosine_similarity` scores how aligned term distributions are (1 = identical direction).
